```python
import pandas as pd
import matplotlib
import transformers
import tokenizers
import sentencepiece

print("Pandas:", pd.__version__)
print("Transformers:", transformers.__version__)
print("Tokenizers:", tokenizers.__version__)
print("SentencePiece: installed")
```

```python
corpus = """
Natural language processing enables computers to understand human language.
Large language models process text using tokens.
Tokenization is an important step in natural language processing.
Subword tokenization helps handle rare and unknown words.
Transformers use powerful tokenization techniques for language understanding.
"""

print(corpus)
```

```python
import re

words = re.findall(r"\b\w+\b", corpus.lower())

print("Total tokens:", len(words))
print("First 20 tokens:", words[:20])
```

```python
word_vocab = sorted(set(words))

print("Word-level vocabulary size:", len(word_vocab))
print("Vocabulary:", word_vocab)
```

```python
from transformers import AutoTokenizer

# BPE tokenizer
gpt2_tokenizer = AutoTokenizer.from_pretrained("gpt2")

# WordPiece tokenizer
bert_tokenizer = AutoTokenizer.from_pretrained("bert-base-uncased")

# SentencePiece tokenizer
t5_tokenizer = AutoTokenizer.from_pretrained("t5-small")

print("GPT-2 BPE tokenizer loaded")
print("BERT WordPiece tokenizer loaded")
print("T5 SentencePiece tokenizer loaded")
```

```python
# Tokenize the corpus using the three subword tokenizers

gpt2_tokens = gpt2_tokenizer.tokenize(corpus)
bert_tokens = bert_tokenizer.tokenize(corpus)
t5_tokens = t5_tokenizer.tokenize(corpus)

print("BPE (GPT-2) tokens:")
print(gpt2_tokens[:50])

print("\nWordPiece (BERT) tokens:")
print(bert_tokens[:50])

print("\nSentencePiece (T5) tokens:")
print(t5_tokens[:50])
```

```python
print("Total BPE tokens:", len(gpt2_tokens))
print("Total WordPiece tokens:", len(bert_tokens))
print("Total SentencePiece tokens:", len(t5_tokens))
```

```python
# Calculate average tokens per word

num_words = len(words)

avg_bpe = len(gpt2_tokens) / num_words
avg_wordpiece = len(bert_tokens) / num_words
avg_sentencepiece = len(t5_tokens) / num_words

print("Average tokens per word:")
print("BPE (GPT-2):", round(avg_bpe, 2))
print("WordPiece (BERT):", round(avg_wordpiece, 2))
print("SentencePiece (T5):", round(avg_sentencepiece, 2))
```

```python
# Vocabulary size observed in our corpus

bpe_vocab = set(gpt2_tokens)
wordpiece_vocab = set(bert_tokens)
sentencepiece_vocab = set(t5_tokens)

print("Vocabulary size observed in corpus:")
print("Word-level:", len(word_vocab))
print("BPE (GPT-2):", len(bpe_vocab))
print("WordPiece (BERT):", len(wordpiece_vocab))
print("SentencePiece (T5):", len(sentencepiece_vocab))
```

```python
import pandas as pd

comparison = pd.DataFrame({
    "Method": [
        "Word-level",
        "BPE (GPT-2)",
        "WordPiece (BERT)",
        "SentencePiece (T5)"
    ],
    "Vocabulary Size": [
        len(word_vocab),
        len(bpe_vocab),
        len(wordpiece_vocab),
        len(sentencepiece_vocab)
    ],
    "Total Tokens": [
        len(words),
        len(gpt2_tokens),
        len(bert_tokens),
        len(t5_tokens)
    ],
    "Average Tokens/Word": [
        1.0,
        round(avg_bpe, 2),
        round(avg_wordpiece, 2),
        round(avg_sentencepiece, 2)
    ]
})

comparison
```

```python
test_words = [
    "tokenization",
    "neuralnetwork",
    "cybersecurity",
    "electromagnetism",
    "computerr"
]

for word in test_words:
    print("\nWord:", word)
    print("Word-level:", [word])
    print("BPE:", gpt2_tokenizer.tokenize(word))
    print("WordPiece:", bert_tokenizer.tokenize(word))
    print("SentencePiece:", t5_tokenizer.tokenize(word))
```

```python
# OOV handling test

oov_words = [
    "computerr",
    "quantumxyz",
    "neuralnetwork"
]

for word in oov_words:
    print("\nWord:", word)

    # Word-level
    if word in word_vocab:
        print("Word-level: Found")
    else:
        print("Word-level: OOV")

    # BPE
    print("BPE:", gpt2_tokenizer.tokenize(word))

    # WordPiece
    print("WordPiece:", bert_tokenizer.tokenize(word))

    # SentencePiece
    print("SentencePiece:", t5_tokenizer.tokenize(word))
```

```python
# Check whether each tokenizer produces an unknown token

for word in oov_words:
    print("\nWord:", word)

    bpe = gpt2_tokenizer.tokenize(word)
    wp = bert_tokenizer.tokenize(word)
    sp = t5_tokenizer.tokenize(word)

    print("BPE:", bpe, "| UNK:", "<unk>" in [x.lower() for x in bpe])
    print("WordPiece:", wp, "| UNK:", "[UNK]" in wp)
    print("SentencePiece:", sp, "| UNK:", "<unk>" in [x.lower() for x in sp])
```

```python
final_comparison = pd.DataFrame({
    "Method": [
        "Word-level",
        "BPE (GPT-2)",
        "WordPiece (BERT)",
        "SentencePiece (T5)"
    ],
    "Vocabulary Size": [
        len(word_vocab),
        len(bpe_vocab),
        len(wordpiece_vocab),
        len(sentencepiece_vocab)
    ],
    "Total Tokens": [
        len(words),
        len(gpt2_tokens),
        len(bert_tokens),
        len(t5_tokens)
    ],
    "Average Tokens/Word": [
        1.00,
        round(avg_bpe, 2),
        round(avg_wordpiece, 2),
        round(avg_sentencepiece, 2)
    ],
    "OOV Strategy": [
        "Whole-word OOV",
        "Subword decomposition",
        "Subword decomposition",
        "Subword decomposition"
    ]
})

final_comparison
```
